```python
model_config = {
    4: {
        "minute_threshold": 1,
        "features": {
            "num_features": [
                "cumsum_minutes",
                "transfers_in_ewm_5",
                "ict_index_ewm_5",
                "selected",
            ],
            "cat_features": [
                "next_week_is_home",
                "players_team",
                "next_week_opponent_team_name",
                "position_name",
            ],
        },
        "preprocessor": {
            "steps": [
                {
                    "name": "num",
                    "transformer": "sklearn.preprocessing.RobustScaler",
                    "columns": [
                        "cumsum_minutes",
                        "transfers_in_ewm_5",
                        "ict_index_ewm_5",
                        "selected",
                    ],
                },
                {
                    "name": "cat",
                    "transformer": "sklearn.preprocessing.OneHotEncoder",
                    "columns": [
                        "next_week_is_home",
                        "players_team",
                        "next_week_opponent_team_name",
                        "position_name",
                    ],
                },
            ]
        },
        "model": {
            "class": "xgboost.XGBRegressor",
            "hyperparams": {
                "booster": "gbtree",
                "colsample_bylevel": 0.8203728667728196,
                "colsample_bytree": 0.922991157077929,
                "gamma": 4.158210875794536,
                "learning_rate": 0.006952873032336879,
                "max_depth": 5,
                "min_child_weight": 1,
                "n_estimators": 773,
                "reg_alpha": 0.008785958264239105,
                "reg_lambda": 5.742011472489027e-07,
                "subsample": 0.8604270021819085,
                "random_state": 42,
            },
        },
    }
}
```

```python
rolling_features = {
            'transfers_in': {'method': 'ewm', 'n': 5},
            'ict_index': {'method': 'ewm', 'n': 5},
        }
```

```python
df = eng_rolling_avg_features(df, rolling_features)
```

```python
pipeline, features = load_config(model_config, model_num=4)
train_df, test_df = train_test_split(df=df, predicting_gameweek=35)
```

```python
model = train_model(train_df, pipeline,features,predicting_gameweek =35, mlflow_tracking_uri=None)
```

```python
y_pred_test, y_true_test, y_pred_train, y_true_train = _get_train_test_predictions(test_df, train_df, features, model, target_col='next_week_round_points')
```

```python
opt_team = _get_optimal_team_test_prediction(test_df,y_pred_test,y_true_test,objective_col_name="predicted_next_week_points")
```

```python
opt_team['squad_ranking']
```

```python
picked_team_gw = opt_team['squad'] 

true_players_stats = df[
            (df['round'] == 35) & 
            (df['player_name'].isin(picked_team_gw))
        ]
predictions = opt_team['squad_ranking'][['player_name', 'predicted_next_week_points', 'rank']]
        
# Merge actual stats with predictions
joined_data = true_players_stats.merge(
    predictions, 
    on='player_name', 
    how='inner'
).sort_values(by='rank')
```

```python
# Aggregate to player level (handles DGWs correctly)
agg_df = (
    joined_data
    .groupby(['player_id'], as_index=False)
    .agg({
        "predicted_next_week_points": 'sum',              # predicted points
        'next_week_round_points': 'sum',         # actual points
        'transfer_cost': 'first',
        'position_name': 'first',
        'players_team': 'first',
        'player_name': 'first'
    })
)
```
